#### Cleaning unit_price ($ strings -> float) & fix invalid/sentinel numbers


```python
cleaned_costumer_transition_data["unit_price"] = (
    cleaned_costumer_transition_data["unit_price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.strip()
)
cleaned_costumer_transition_data["unit_price"] = pd.to_numeric(cleaned_costumer_transition_data["unit_price"], errors="coerce")

cleaned_costumer_transition_data.loc[(cleaned_costumer_transition_data["age"] < 0) | (cleaned_costumer_transition_data["age"] > 110), "age"] = np.nan
cleaned_costumer_transition_data.loc[cleaned_costumer_transition_data["satisfaction_rating"] == 99, "satisfaction_rating"] = np.nan
```

```python
cleaned_costumer_transition_data.info()
```

#### Filling missing numbers with mediann  cast integer columns to int

```python
int_cols = ["age", "quantity", "loyalty_points", "satisfaction_rating"]
for col in int_cols:
    cleaned_costumer_transition_data[col] = cleaned_costumer_transition_data[col].fillna(cleaned_costumer_transition_data[col].median()).round().astype(int)

cleaned_costumer_transition_data["unit_price"] = cleaned_costumer_transition_data["unit_price"].fillna(cleaned_costumer_transition_data["unit_price"].median())
```

#### Cleaning  filling string columns

```python
cleaned_costumer_transition_data["full_name"] = cleaned_costumer_transition_data["full_name"].fillna("Unknown").str.strip().str.title()

cleaned_costumer_transition_data["city"] = (
    cleaned_costumer_transition_data["city"]
    .replace(["N/A", "unknown", ""], np.nan)
    .fillna("Unknown")
    .str.strip()
    .str.title()
    .replace({"La": "Los Angeles"})
)

cleaned_costumer_transition_data["product_category"] = (
   cleaned_costumer_transition_data["product_category"]
    .fillna("Unknown")
    .str.strip()
    .str.title()
)
```

```python
cleaned_costumer_transition_data.head()
```

#### Encoding -> Standardize is_active booleans  cast to int (0 or 1)

```python
active_map = {
    "Yes": 1, "yes": 1, "1": 1, "True": 1, "true": 1,
    "No": 0, "no": 0, "0": 0, "False": 0, "false": 0
}
cleaned_costumer_transition_data["is_active"] = cleaned_costumer_transition_data["is_active"].map(active_map)
cleaned_costumer_transition_data["is_active"] = cleaned_costumer_transition_data["is_active"].fillna(cleaned_costumer_transition_data["is_active"].mode()[0]).astype(int)
```

#### Parse mixed signup_date formats  fill missing dates

```python
cleaned_costumer_transition_data["signup_date"] = pd.to_datetime(cleaned_costumer_transition_data["signup_date"], format="mixed", errors="coerce")
cleaned_costumer_transition_data["signup_date"] = cleaned_costumer_transition_data["signup_date"].fillna(cleaned_costumer_transition_data["signup_date"].median())
```
