## cleaning data


##### checking duplicates

```python
duplicated_values=video_games_sales[video_games_sales.duplicated()]
duplicated_values
```


 Removeing duplicates

```python
video_games_sales=video_games_sales.drop_duplicates()
```

##### filling null value and cleaning


```python
video_games_sales.isnull()
```

```python
video_games_sales.isnull().sum()
```

```python
video_games_sales=video_games_sales.fillna('NOT FOUND')
```

removeing elements where object is not int

```python
video_games_sales['NA_Sales'] = video_games_sales['NA_Sales'].replace('[$]', '', regex=True,)
```

```python
video_games_sales.head(20)
```

average sales

```python
average_sales=video_games_sales['NA_Sales'].mean()
```

```python
average_sales=int(average_sales)
```

```python
video_games_sales['NA_Sales']=average_sales=video_games_sales['NA_Sales'].fillna(average_sales)
```

### modifying the country column

```python
video_games_sales['Country']=video_games_sales['Country'].replace({'USA':'United States'})
```

```python
video_games_sales['Country']=video_games_sales['Country'].str.title()
```

#### selecting

```python
# United states sales 

united_states=video_games_sales[video_games_sales['Country']=='United States']
united_states
```

```python
# Australia_sales

Australia_sales=video_games_sales[video_games_sales['Country']=='Australia']
Australia_sales
```

## Renameing columns

```python
video_games_sales=video_games_sales.rename(columns={'NA_Sales':'National sales','NA_Profit':'National profit'})
video_games_sales
```

  ###  winsorizing

```python
sales_cap = video_games_sales['National sales'].quantile(0.95)
sales_cap
video_games_sales['National Sales'] = np.where(video_games_sales['Global_Sales'] > sales_cap, sales_cap, video_games_sales['Global_Sales'])
```

```python
video_games_sales.head(10)
```

# DATA VISUALIZATION

```python

video_games_sales['NA_Sales'] = pd.to_numeric(video_games_sales['NA_Sales'], errors='coerce')

national_Sales = video_games_sales.groupby(['Region', 'Country'])['NA_Sales'].sum().reset_index()
```

## creating a barchart


```python
plt.figure(figsize=(16,8))
sns.barplot(data=national_Sales,x='Region',y='NA_Sales',hue='Country')
plt.title('National_Sales by region and country',fontweight='bold')
plt.xlabel(' region ',fontweight='bold')
plt.ylabel('National_Sales',fontweight='bold')
plt.show()
```

## creating a box_plot

```python

plt.figure(figsize=(16, 8))
sns.boxplot(data=video_games_sales, x='Region', y='NA_Sales', hue='Country')

plt.title('Distribution of NA Sales by Region and Country', fontsize=16)
plt.xticks(rotation=45)
plt.show()
```

### create a piechart

```python

pie_data = video_games_sales.groupby('Country')['NA_Sales'].sum()


plt.figure(figsize=(10, 8))
plt.pie(
    pie_data,
    labels=pie_data.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('pastel'),
    wedgeprops={'edgecolor': 'black'},
)

plt.title('NA Sales Share by Country', fontsize=16, fontweight='bold')
plt.show()
```
