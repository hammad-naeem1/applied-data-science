#### Distrubution plot

```python
numeric_columns = insurance_data.select_dtypes(include=['number']).columns.tolist()

for col in numeric_columns:
    plt.figure(figsize=(6, 4))
    sns.histplot(insurance_data[col], kde=True, bins=20)
    plt.title(f'Distribution of {col}')
    plt.show()
```

#### Count plot

```python
# counting sex
sns.countplot(x=insurance_data['sex'])
```

```python
# counting smokers
sns.countplot(x=insurance_data['smoker'])
```

#### Box plot

```python
for col in numeric_columns:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=insurance_data[col])
    plt.title(f'Box plot of {col}')
    plt.show()
```

## Data Cleaning and data preprocessing

```python
cleaned_insurance_data=insurance_data.copy()
```

```python
cleaned_insurance_data.head()
```

### Removing duplicates

```python
cleaned_insurance_data.drop_duplicates(inplace=True)
```

### Removing null or missing values

```python
cleaned_insurance_data.isna()
```

```python
cleaned_insurance_data.isna().sum()
```

### Converting datatypes to int

```python
cleaned_insurance_data
```

### Label Encoding

##### converting sex column

```python
cleaned_insurance_data['sex'].value_counts()
```

```python
cleaned_insurance_data['sex']=cleaned_insurance_data['sex'].map({'male':0 ,'female':1})
```

##### converting smoker column


```python
cleaned_insurance_data['smoker'].value_counts()
```

```python
cleaned_insurance_data['smoker']=cleaned_insurance_data['smoker'].map({'no':0 ,'yes':1})
```

### one-hot encoding

```python
cleaned_insurance_data['region'].value_counts()
```

#### converting Region's column

```python
cleaned_insurance_data=pd.get_dummies(cleaned_insurance_data,columns=['region'],drop_first=True)
```

```python
cleaned_insurance_data
```

```python
cleaned_insurance_data=cleaned_insurance_data.astype(int)
```

```python
cleaned_insurance_data
```

### changing column's names

```python
cleaned_insurance_data.rename(columns={
    'smoker':'is_smoker',
    'sex':'is_female'
},inplace=True)
```
