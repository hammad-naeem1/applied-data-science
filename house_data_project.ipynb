## Data-Cleaning

```python
cleaned_house_data=data.copy()
```

Removing Duplicates


```python
cleaned_house_data.duplicated().sum()
```

```python
cleaned_house_data.drop_duplicates(inplace=True)
```

Removing null or missing values


```python
cleaned_house_data.isna()
```

```python
cleaned_house_data['date'].fillna('not-found',inplace=True)
```

```python
cleaned_house_data.select_dtypes(include=['int', 'float']).fillna(0,inplace=True)
```

```python
cleaned_house_data.head()
```

Removing columns

```python
cleaned_house_data.drop(columns=['yr_renovated','sqft_basement','zipcode','waterfront','long','view','condition',
                                'grade' ],inplace=True)
```

## Data-visualization

```python
cleaned_house_data
```

Barplot

```python
sns.barplot(x=cleaned_house_data['bedrooms'],y=cleaned_house_data['price'],palette='rainbow')
```

countplot

```python
cleaned_house_data.head()
```

```python
# counting bedrooms
sns.countplot(x=cleaned_house_data['bedrooms'],palette='rocket')
```

```python
# counting bathrooms

sns.countplot(y=cleaned_house_data['bathrooms'],palette='flare')
```

```python
# counting floors

sns.countplot(x=cleaned_house_data['floors'],palette='magma')
```

Barplot

```python
numeric_columns = cleaned_house_data.select_dtypes(include=['number']).columns.tolist()
```

```python
for col in numeric_columns:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=cleaned_house_data[col])
    plt.title(f'Box plot of {col}')
    plt.show()
```

Distrubution plot

```python
for col in numeric_columns:
    plt.figure(figsize=(6, 4))
    sns.histplot(cleaned_house_data[col], kde=True, bins=20)
    plt.title(f'Distribution of {col}')
    plt.show()
```
